se eligio las proyecciones de la poblacion  del municipio de San lucas Sacatepequez y sus datos de luminocidad nocturna y metros cuadrados construidos , tomando como varible indepediente la proyeccion de poblacion y los otros datos como dependientes tanto la luminocidad como los metros cuadrados construidos explorando su posible relacion con la evolucion democgrafica  





Demostracion dinamica urbana municipio de prueba:
poblacion INE PROYECTADA VS SUPERFICIE CONSTRUIDA
Municipio de San Lucas Sacatepequez fuentes
Open Building y proyecciones municipales INE

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, linregress
```

municipio san lucas sacatepequez

```python

municipio = "Municipio de ejemplo"

anios = np.array([
     2016, 2017, 2018, 2019,
    2020, 2021, 2022, 2023,
])

poblacion = np.array([
  26840, 27744, 27681, 28066,
    28445, 28819, 29187, 29549,
])

superficie_km2 = np.array([
    1.067, 1.161, 1.197, 1.244, 1.317,
    1.414, 1.456, 1.582,
])
```

Correlacion de pearson

```python

r, p = pearsonr(poblacion, superficie_km2)

print("Municipio:", municipio)
print("Correlación de Pearson:", round(r, 3))
```

linea de tendencia

```python

regresion = linregress(poblacion, superficie_km2)

pendiente = regresion.slope
intercepto = regresion.intercept

# Valores para dibujar la línea
x_tendencia = np.linspace(
    poblacion.min(),
    poblacion.max(),
    100
)

y_tendencia = (
    pendiente * x_tendencia + intercepto
)

print(
    "Ecuación de tendencia:",
    f"Superficie = {pendiente:.6f} * Población + {intercepto:.3f}"
)
```

grafica

```python

plt.figure(figsize=(9, 6))

plt.scatter(
    poblacion,
    superficie_km2,
    label="Datos anuales"
)

plt.plot(
    x_tendencia,
    y_tendencia,
    label="Línea de tendencia"
)

plt.title(
    f"{municipio}\nPoblación proyectada y superficie construida"
)

plt.xlabel("Población proyectada (INE)")
plt.ylabel("Superficie construida (km²)")

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
